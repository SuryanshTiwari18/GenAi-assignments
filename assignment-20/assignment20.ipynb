## Task 2: Text Preprocessing for Recommendation

In this task, the textual information of each book will be cleaned before vectorization.

The following preprocessing steps will be performed:

- Replace missing values with empty strings
- Combine important text features into one column
- Convert text to lowercase
- Remove punctuation and special characters
- Remove stopwords
- Store the processed text in a new column named **clean_text**

```python
import re
import string
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
text_columns=["title","authors","categories","description"]
df[text_columns] = df[text_columns].fillna("")
```

```python
df["combined_text"]=(df["title"] + " " + df["authors"] + " " + df["categories"] + " " + df["description"])
```

```python
stop_words=set(stopwords.words("english"))
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return " ".join(words)
df["clean_text"] = df["combined_text"].apply(clean_text)
```

```python
df[["title", "clean_text"]].head()
```

```python
print("Missing values in clean_text:")
print(df["clean_text"].isnull().sum())
print("\nEmpty clean_text rows:")
print((df["clean_text"] == "").sum())
```

# Task 3: TF-IDF Vectorization

Machine learning models cannot directly understand textual data. Therefore, the cleaned text must be converted into numerical feature vectors.

In this task, **TF-IDF (Term Frequency–Inverse Document Frequency)** is used to transform the `clean_text` column into a numerical matrix.

TF-IDF assigns higher weights to words that are important in a document while reducing the importance of very common words that appear across many documents.

This TF-IDF matrix will be used in the next task to calculate the similarity between books using **Cosine Similarity**.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(max_features=5000,stop_words="english")
tfidf_matrix=tfidf.fit_transform(df["clean_text"])
```

```python
print("TF-IDf Matrix Shape:", tfidf_matrix.shape)
print("Vocabulary Size:", len(tfidf.get_feature_names_out()))
```

```python
vocabulary = tfidf.get_feature_names_out()
print("First 30 words in the vocabulary:", vocabulary[:30])
```

```python
tfidf_df=pd.DataFrame(tfidf_matrix[:5].toarray(), columns=vocabulary)
tfidf_df.iloc[:, :20]
```

```python
first_book=tfidf_df.iloc[0]
top_words=(first_book[first_book>0].sort_values(ascending=False).head(10))
top_words
```

### Observations

- The TF-IDF vectorizer successfully transformed the cleaned textual data into numerical feature vectors.
- The resulting matrix has **6810 rows** (books) and **5000 columns** (selected vocabulary terms).
- Each book is represented as a 5000-dimensional vector.
- Words with higher TF-IDF scores are considered more representative of a particular book, while very common words receive lower weights.
- These vectors will be used in the next task to compute similarity between books using Cosine Similarity.

# Task 4: Computing Cosine Similarity

After converting the textual information into TF-IDF vectors, the next step is to measure the similarity between books.

Cosine Similarity is one of the most commonly used similarity measures in content-based recommendation systems. It calculates the cosine of the angle between two TF-IDF vectors.

- A similarity score of **1** indicates that two books are identical in terms of their textual representation.
- A similarity score of **0** indicates that the books are completely different.

The resulting similarity matrix will later be used to recommend books that are most similar to a selected book.

```python
from sklearn.metrics.pairwise import cosine_similarity
cosine_sim = cosine_similarity(tfidf_matrix)
```

```python
print("Cosine Similarity Matrix Shape:", cosine_sim.shape)
```

```python
similarity_df=pd.DataFrame(cosine_sim[:5, :5],index=df["title"][:5],columns=df["title"][:5])
similarity_df
```

```python
print("similarity between Book 1 and Book 2:", cosine_sim[0][1])
```

```python
similar_scores = list(enumerate(cosine_sim[0]))

similar_scores = sorted(
    similar_scores,
    key=lambda x: x[1],
    reverse=True
)

similar_scores[:10]
```

### Observations

- Cosine Similarity successfully computed the similarity between every pair of books.
- The similarity matrix has dimensions **6810 × 6810**, where each row and column corresponds to a book in the dataset.
- Every diagonal value is equal to **1**, indicating that each book is perfectly similar to itself.
- Higher similarity scores indicate that two books share more similar textual content based on their TF-IDF representations.
- This similarity matrix forms the foundation of the recommendation engine and will be used to retrieve the most relevant books in the next task.

```python
indices = pd.Series(df.index, index=df["title"]).drop_duplicates()
indices.head()
```

# Task 5: Building the Book Recommendation Function

In this task, a recommendation function is created using the cosine similarity matrix.

The function performs the following steps:

- Accepts a book title as input.
- Finds the corresponding index of the selected book.
- Retrieves the similarity scores of all books.
- Sorts the books based on similarity scores.
- Excludes the selected book itself.
- Returns the top recommended books along with their similarity scores.

```python
def recommend_books(title, top_n=5):
    if title not in indices:
        print("Book not found in the dataset.")
        return

    idx = indices[title]
    similarity_scores = list(enumerate(cosine_sim[idx]))
    similarity_scores = sorted(similarity_scores,key=lambda x: x[1], reverse=True)
    similarity_scores = similarity_scores[1:top_n+1]
    recommendations = []
    for i,score in similarity_scores:
        recommendations.append({
            "Title": df.iloc[i]["title"],
            "Author": df.iloc[i]["authors"],
            "Category": df.iloc[i]["categories"],
            "Similarity Score": round(score, 3)
        })
    return pd.DataFrame(recommendations)
```

```python
recommend_books("Gilead")
```
